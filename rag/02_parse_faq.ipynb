```python
# Cell 1 — Setup
catalog  = "bootcamp_students"
schema   = "maintops"
print(f"Working in: {catalog}.{schema}")
```

```python
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {catalog}.{schema}.raw_faq (
  doc_id            STRING    COMMENT 'Unique document identifier (UUID)',
  source_type       STRING    COMMENT 'pdf | web | image',
  source_url        STRING    COMMENT 'File path or URL the content was extracted from',
  title             STRING    COMMENT 'Human-readable document title',
  content           STRING    COMMENT 'Extracted plain text content',
  ingested_at       TIMESTAMP COMMENT 'Ingestion timestamp'
)
TBLPROPERTIES (delta.enableChangeDataFeed = true)
"""
)
```

```python
PDF_FILENAME   = "MaintOps_Public_User_Guide.pdf"
VOL_PATH       = f"/Volumes/{catalog}/{schema}/maintops_docs/"
PDF_VOL_PATH   = f"{VOL_PATH}{PDF_FILENAME}"
IMAGE_DIR      = f"{VOL_PATH}faq_images/"   # images extracted by ai_parse_document
```

```python
# Parse the FAQ PDF, fail loudly on parser errors or empty output, then upsert by file path.
# The parsed rows are collected to the driver so ai_parse_document runs only once (a handful of documents).
parsed = spark.sql(f"""
WITH parsed_documents AS (
  SELECT
    regexp_replace(path, '^dbfs:', '') AS source_url,
    ai_parse_document(
      content,
      map(
        'imageOutputPath', '{IMAGE_DIR}',
        'descriptionElementTypes', '*'
      )
    ) AS parsed
  FROM READ_FILES('{PDF_VOL_PATH}', format => 'binaryFile')
)
SELECT
  sha2(source_url, 256)                                   AS doc_id,
  'pdf'                                                   AS source_type,
  source_url,
  'Maintops FAQ'                                          AS title,
  concat_ws(
    '\n\n',
    transform(
      try_cast(parsed:document:elements AS ARRAY<VARIANT>),
      element -> try_cast(element:content AS STRING)
    )
  )                                                       AS content,
  try_cast(parsed:error_status AS STRING)                 AS error_status,
  current_timestamp()                                     AS ingested_at
FROM parsed_documents
""").collect()

errors = [r.source_url for r in parsed if r.error_status is not None]
empty  = [r.source_url for r in parsed if r.error_status is None and not (r.content or "").strip()]
if not parsed:
    raise RuntimeError(f"No documents found at {PDF_VOL_PATH}")
if errors or empty:
    raise RuntimeError(f"Parsing failed for {errors}, empty text for {empty}")

# error_status is all NULL here, so leave it out (Spark can't infer the type of an all-NULL column)
(spark.createDataFrame([{k: v for k, v in r.asDict().items() if k != "error_status"} for r in parsed])
      .createOrReplaceTempView("parsed_faq"))

spark.sql(f"""
MERGE INTO {catalog}.{schema}.raw_faq AS t
USING parsed_faq AS s
ON t.source_url = s.source_url
WHEN MATCHED AND t.content <> s.content THEN UPDATE SET *
WHEN NOT MATCHED THEN INSERT *
""")

display(spark.table(f"{catalog}.{schema}.raw_faq"))
```
