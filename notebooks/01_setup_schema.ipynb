```python
# Cell 1 — Identify your username and set catalog/schema names
from databricks.sdk import WorkspaceClient

w        = WorkspaceClient()
me       = w.current_user.me()
username = me.user_name.split("@")[0].replace(".", "_").replace("-", "_")

catalog = "bootcamp_students"
schema  = f"maintops"

print(f"Catalog : {catalog}")
print(f"Schema  : {schema}")
print(f"User    : {me.user_name}")
```

```python
# Cell 2 — Create personal schema
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.{schema}")
print(f"Schema ready: {catalog}.{schema}")
```

```python
spark.sql(f"CREATE VOLUME IF NOT EXISTS {catalog}.{schema}.maintops_docs")
print(f"Volume ready: /Volumes/{catalog}/{schema}/faq/")
```

```python
display(spark.sql(f"SHOW VOLUMES IN {catalog}.{schema}"))
```

```python
statements = [
    # f"GRANT USAGE ON CATALOG {catalog} TO `account users`",
    f"GRANT USAGE ON SCHEMA {catalog}.{schema} TO `account users`",
    f"GRANT READ VOLUME ON VOLUME {catalog}.{schema}.maintops_docs TO `account users`",
    f"GRANT WRITE VOLUME ON VOLUME {catalog}.{schema}.maintops_docs TO `account users`",
]
for stmt in statements:
    try:
        spark.sql(stmt)
    except Exception as e:
        # `account users` may not exist in every workspace — don't fail the whole notebook over it
        print(f"Skipped grant (non-fatal): {stmt}\n  {e}")

display(spark.sql(f"SHOW GRANTS ON VOLUME {catalog}.{schema}.maintops_docs"))
```
